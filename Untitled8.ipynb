```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy import stats

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: f'{x:,.2f}')
```

```python
from google.colab import files

uploaded = files.upload()

csv_name = next((name for name in uploaded.keys() if name.lower().endswith('.csv')), None)

if csv_name is None:
    raise ValueError("Please upload a CSV file.")

print("Using file:", csv_name)
```

```python
df = pd.read_csv(csv_name)

print("Dataset shape:", df.shape)
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame("dtype"))

print("\nMissing values:")
display(df.isnull().sum().to_frame("missing_values"))

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
data = df.copy()

# Remove exact duplicate rows if any
data = data.drop_duplicates().copy()

missing_cols = ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']

for col in missing_cols:
    data[col] = data.groupby('Season')[col].transform(
        lambda x: x.fillna(x.median())
    )

print("Missing values after cleaning:")
display(data.isnull().sum().to_frame("missing_values"))

print("Cleaned dataset shape:", data.shape)
```

```python
print("Number of farms:", data['Farm_ID'].nunique())
print("States:", data['State'].nunique())
print("Districts:", data['District'].nunique())
print("Crops:", data['Crop'].nunique())

print("\nSeason distribution:")
display(data['Season'].value_counts().to_frame("records"))

print("\nCrop distribution:")
display(data['Crop'].value_counts().to_frame("records"))

print("\nIrrigation methods:")
display(data['Irrigation_Method'].value_counts().to_frame("records"))
```

```python
performance_cols = [
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Revenue_INR',
    'Profit_INR',
    'Water_Efficiency_t_per_1000m3'
]

season_performance = data.groupby('Season')[performance_cols].mean().round(2)
display(season_performance)

fig, ax = plt.subplots(figsize=(8,5))
season_performance['Yield_Tonnes_Ha'].plot(kind='bar', ax=ax)
ax.set_title('Average Yield by Season')
ax.set_ylabel('Yield (Tonnes/Ha)')
ax.set_xlabel('Season')
plt.xticks(rotation=0)
plt.show()

fig, ax = plt.subplots(figsize=(8,5))
season_performance['Profit_INR'].plot(kind='bar', ax=ax)
ax.set_title('Average Profit by Season')
ax.set_ylabel('Average Profit (INR)')
ax.set_xlabel('Season')
plt.xticks(rotation=0)
plt.show()
```

```python
season_environment = data.groupby('Season')[
    ['Rainfall_mm','Avg_Temperature_C','Humidity_pct',
     'Sunlight_Hours_Day','Soil_Moisture_pct','Disease_Pest_Risk_pct']
].mean().round(2)

display(season_environment)

for col in season_environment.columns:
    fig, ax = plt.subplots(figsize=(7,4))
    season_environment[col].plot(kind='bar', ax=ax)
    ax.set_title(f'Average {col} by Season')
    ax.set_xlabel('Season')
    ax.set_ylabel(col)
    plt.xticks(rotation=0)
    plt.show()
```

```python
characteristics = [
    'Rainfall_mm','Avg_Temperature_C','Humidity_pct','Sunlight_Hours_Day',
    'Soil_Moisture_pct','Disease_Pest_Risk_pct','Yield_Tonnes_Ha',
    'Profit_INR','Water_Efficiency_t_per_1000m3'
]

display(data.groupby('Season')[characteristics].mean().round(2).T)
```

```python
crop_share = pd.crosstab(data['Season'], data['Crop'], normalize='index') * 100
irrigation_share = pd.crosstab(data['Season'], data['Irrigation_Method'], normalize='index') * 100

print("Crop share within each season (%)")
display(crop_share.round(2))

print("Irrigation-method share within each season (%)")
display(irrigation_share.round(2))
```

```python
resource_cols = [
    'Fertilizer_kg_ha','Pesticide_Litre_ha',
    'Water_Used_m3','Water_Efficiency_t_per_1000m3',
    'Nitrogen_kg_ha','Phosphorus_kg_ha','Potassium_kg_ha'
]

resource_summary = data.groupby('Season')[resource_cols].mean().round(2)
display(resource_summary)

fig, ax = plt.subplots(figsize=(8,5))
resource_summary['Water_Used_m3'].plot(kind='bar', ax=ax)
ax.set_title('Average Water Used by Season')
ax.set_ylabel('Water Used (m³)')
ax.set_xlabel('Season')
plt.xticks(rotation=0)
plt.show()

fig, ax = plt.subplots(figsize=(8,5))
resource_summary['Water_Efficiency_t_per_1000m3'].plot(kind='bar', ax=ax)
ax.set_title('Average Water Efficiency by Season')
ax.set_ylabel('Tonnes per 1000 m³')
ax.set_xlabel('Season')
plt.xticks(rotation=0)
plt.show()
```

```python
environment_cols = [
    'Rainfall_mm','Avg_Temperature_C','Humidity_pct',
    'Sunlight_Hours_Day','Soil_Moisture_pct','Disease_Pest_Risk_pct'
]

performance_cols = [
    'Yield_Tonnes_Ha','Production_Tonnes','Profit_INR',
    'Revenue_INR','Water_Efficiency_t_per_1000m3'
]

correlation_table = data[environment_cols + performance_cols].corr().loc[
    environment_cols, performance_cols
].round(3)

display(correlation_table)

fig, ax = plt.subplots(figsize=(9,5))
im = ax.imshow(correlation_table.values, aspect='auto')
ax.set_xticks(range(len(correlation_table.columns)))
ax.set_xticklabels(correlation_table.columns, rotation=45, ha='right')
ax.set_yticks(range(len(correlation_table.index)))
ax.set_yticklabels(correlation_table.index)
ax.set_title('Environmental Variables vs Performance Correlations')
plt.colorbar(im, ax=ax)
plt.tight_layout()
plt.show()
```

```python
economic_summary = data.groupby('Season').agg(
    Average_Market_Price=('Market_Price_INR_Tonne','mean'),
    Average_Cost=('Total_Cost_INR','mean'),
    Average_Revenue=('Revenue_INR','mean'),
    Average_Profit=('Profit_INR','mean'),
    Median_Profit=('Profit_INR','median'),
    Profitable_Farms_Pct=('Profit_INR', lambda x: (x > 0).mean() * 100)
).round(2)

display(economic_summary)
```

```python
state_profit = data.pivot_table(
    index='State', columns='Season', values='Profit_INR', aggfunc='mean'
).round(2)

crop_profit = data.pivot_table(
    index='Crop', columns='Season', values='Profit_INR', aggfunc='mean'
).round(2)

print("Average profit by State and Season")
display(state_profit)

print("Average profit by Crop and Season")
display(crop_profit)

# Check Kharif > Rabi > Zaid
if all(s in state_profit.columns for s in ['Kharif','Rabi','Zaid']):
    state_order = (
        (state_profit['Kharif'] > state_profit['Rabi']) &
        (state_profit['Rabi'] > state_profit['Zaid'])
    ).sum()
    print(f"States with Kharif > Rabi > Zaid average profit: {state_order} of {len(state_profit)}")

if all(s in crop_profit.columns for s in ['Kharif','Rabi','Zaid']):
    crop_order = (
        (crop_profit['Kharif'] > crop_profit['Rabi']) &
        (crop_profit['Rabi'] > crop_profit['Zaid'])
    ).sum()
    print(f"Crops with Kharif > Rabi > Zaid average profit: {crop_order} of {len(crop_profit)}")
```

```python
state_yield = data.pivot_table(
    index='State', columns='Season', values='Yield_Tonnes_Ha', aggfunc='mean'
).round(2)

display(state_yield)

print("Top 10 highest-profit observations:")
display(data.nlargest(10, 'Profit_INR')[
    ['Farm_ID','State','Crop','Season','Yield_Tonnes_Ha','Revenue_INR','Profit_INR']
])

print("Top 10 largest losses:")
display(data.nsmallest(10, 'Profit_INR')[
    ['Farm_ID','State','Crop','Season','Yield_Tonnes_Ha','Revenue_INR','Profit_INR']
])

# IQR outlier counts
for col in ['Yield_Tonnes_Ha','Profit_INR','Revenue_INR','Water_Used_m3']:
    q1 = data[col].quantile(0.25)
    q3 = data[col].quantile(0.75)
    iqr = q3 - q1
    low = q1 - 1.5 * iqr
    high = q3 + 1.5 * iqr
    outliers = data[(data[col] < low) | (data[col] > high)]
    print(f"{col}: {len(outliers)} IQR outliers")
```

```python
test_cols = [
    'Yield_Tonnes_Ha','Production_Tonnes','Profit_INR','Revenue_INR',
    'Water_Used_m3','Water_Efficiency_t_per_1000m3',
    'Rainfall_mm','Avg_Temperature_C','Humidity_pct',
    'Soil_Moisture_pct','Disease_Pest_Risk_pct'
]

anova_results = []

for col in test_cols:
    groups = [
        group[col].dropna().values
        for _, group in data.groupby('Season')
    ]
    f_stat, p_value = stats.f_oneway(*groups)
    anova_results.append({
        'Variable': col,
        'F_statistic': f_stat,
        'p_value': p_value,
        'Significant_at_0.05': p_value < 0.05
    })

anova_df = pd.DataFrame(anova_results)
display(anova_df)
```
